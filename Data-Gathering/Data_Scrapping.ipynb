# **Comments**

```python
comments = soup.find_all(string=lambda text: isinstance(text, Comment))

for comment in comments:
    print(comment.strip())
```

# **Collecting all texts from Html**

```python
div = soup.find_all('div')
div
```

```python
for l in div:
    print(l.text)
    print(l.get('class')) # getting the names of classes used in every div.
```
