```python
import pandas as pd
from sqlalchemy import create_engine

# SQLAlchemy ইঞ্জিন তৈরি করা (ইউজারনেম, পাসওয়ার্ড, হোস্ট এবং ডেটাবেজের নাম ঠিকমতো বসিয়ে দেবেন)
# যদি পাসওয়ার্ড না থাকে, তবে password অংশটি ফাঁকা বা বাদ দিতে পারেন
engine = create_engine("mysql+pymysql://root:@localhost/world")

# এখন pandas দিয়ে কুয়েরি চালানো (কোনো ওয়ার্নিং ছাড়াই কাজ করবে)
df = pd.read_sql_query("SELECT * FROM city WHERE name LIKE 'Kabul'", engine)

df
```
