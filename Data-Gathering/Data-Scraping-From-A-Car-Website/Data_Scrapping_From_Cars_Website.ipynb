```python
c=0
all_cars_data = []
for i in range(17):
    web = requests.get(f'https://webscraper.io/test-sites/pagination?page=${i}',verify=False)
    soup = BeautifulSoup(web.text,"html.parser")
    boxes = soup.find_all("div",class_="col-md-4 col-xl-4 col-lg-4")
    for box in boxes:
        if box.find("div",class_="badge").text.strip()[0:9] == 'Available':
                stat = box.find("div",class_="badge").text.strip()[0:9]
                num = box.find("div",class_="badge").text.strip()[-1]
                
        else:
            num = 0
            stat = box.find("div",class_="badge").text.strip()
            
        about = box.find("a",class_="card-head-url").get('href')
        
        car_info = box.find_all("a")[1].text.strip()
        
        parts = car_info.split()

        
        brand = parts[0] 
        model = " ".join(parts[1:-1]) 
        year = parts[-1]   
        
            
        
        origin = box.find_all("p",class_="card-text")[1].text.strip()[20:]
        mileage = box.find_all("p",class_="card-text")[2].text.strip()[10:]
        rating = str(len(box.find_all("span",class_="ws-icon ws-icon-star active"))) + " star"
        price = box.find("span",itemprop="price").text[4:]

        details = {
            "stat" : stat,
            "quantity" : num,
            "about" : about,
            "brand" : brand,
            "model" : model,
            "year" : year,
            "origin" : origin,
            "mileage" : mileage,
            "rating" : rating,
            "price" : price,    
        }

        all_cars_data.append(details)
        c=c+1
print(c)
```

```python
all_cars_data
```

```python
import pandas as pd

# ১. Pandas DataFrame-এ রূপান্তর
df = pd.DataFrame(all_cars_data)

# জুপিটার নোটবুকে সুন্দর টেবিল আকারে দেখার জন্য:
display(df.head())
```

```python
# ২. CSV ফাইলে সেভ করা
df.to_csv("car_market_data.csv", index=False)
print("-> CSV file saved successfully!")
```
